```python
import pandas as pd
import glob

path = "/Users/dexter/Documents/Projects/SlangShift/CommonCrawlDiff/data/lang_quality_filtered_contexts/Run0/word_context_*.csv"
all_files = glob.glob(path)
df = pd.concat((pd.read_csv(f) for f in all_files), ignore_index=True)
```

```python
target = 'delulu'
df[target] = df['uri'].apply(lambda x: 1 if target in x else 0)
df[df[target]==1].sample(random_state=473, n=50).to_csv(f"{target}_annotations.csv")
```

```python
df['uri_root'] = df['uri'].apply(lambda x:  x.split('.')[1] if 'www' in x else x.split('.')[0])
df.groupby('uri_root').count().sort_values(by='uri', ascending=False).head(10)
```
