```python
import pandas as pd
import glob

# Mac path
#path = "/Users/dexter/Documents/Projects/SlangShift/CommonCrawlDiff/data/lang_quality_filtered_contexts/Run0/word_context_*.csv"
# Windows path
path = "C:\Projects\CommonCrawlAnalysis\FineWebAnalysis\data\sample-10BT\word_context_*.csv"

all_files = glob.glob(path)
df = pd.concat((pd.read_csv(f) for f in all_files), ignore_index=True)
```

```python
# Spacing is added to some of these for cases when unspaced version show up frequently as substrings.
targets = ['delulu', 'giving', 'alpha', 'chopped', 'epic', 'fire', ' ick ', ' lit ', ' meh ', ' mid ', ' rad ', ' rizz ', 'savage', ' sick ', ' sus ', ' troll ', 'vibes', 'washed']
for target in targets:
    df[target] = df['target_context'].apply(lambda x: 1 if target in x else 0)
    if len(df[df[target]==1]) >= 50:
        df[df[target]==1].sample(random_state=473, n=50).to_csv(f"annotations/fine_web_small_annotations/{target}_annotations.csv", columns=["uri", "target_context"])
```

```python
df['uri_root'] = df['uri'].apply(lambda x:  x.split('.')[1] if 'www' in x else x.split('.')[0])
df.groupby('uri_root').count().sort_values(by='uri', ascending=False).head(10)
```
